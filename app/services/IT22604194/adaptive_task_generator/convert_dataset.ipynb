```python
import json

input_path = r"C:\Users\Dell\OptiCode\adaptive_task_generator\training_data.jsonl"
output_path = r"C:\Users\Dell\OptiCode\adaptive_task_generator\fine_tune_ready.jsonl"

def map_skill(diff):
    if diff <= 2:
        return "Beginner"
    elif diff <= 4:
        return "Intermediate"
    else:
        return "Advanced"

output_file = open(output_path, "w", encoding="utf-8")

with open(input_path, "r", encoding="utf-8") as f:
    for line in f:
        row = json.loads(line)

        skill = map_skill(row["difficulty"])

        instruction = f"Generate a Python coding task for skill level '{skill}'."

        input_text = (
            "Reference problem:\n" +
            row["question"] +
            f"\n\nDifficulty: {row['difficulty']}\n"
        )

        output_data = {
            "title": row["question"][:50],
            "description": row["question"],
            "difficulty": skill,
            "starter_code": ""
        }

        output_file.write(json.dumps({
            "instruction": instruction,
            "input": input_text,
            "output": output_data
        }) + "\n")

output_file.close()

print("Saved:", output_path)
```

```python
def map_difficulty_to_skill(d):
    if d <= 2:
        return "Beginner"
    elif d <= 4:
        return "Intermediate"
    else:
        return "Advanced"

df_filtered['skill_level'] = df_filtered['difficulty'].apply(map_difficulty_to_skill)
```

```python
df_filtered['starter_code'] = df_filtered['starter_code'].fillna('')
```

```python
import json

output_file = "training_data.jsonl"

with open(output_file, 'w', encoding='utf-8') as f:
    for _, row in df_filtered.iterrows():
        instruction = f"Generate a Python coding task for skill level '{row['skill_level']}'"
        input_text = f"Reference problem:\n{row['question']}\nDifficulty: {row['difficulty']}\nStarter code:\n{row['starter_code']}"
        output_data = {
            "title": row['question'][:50],  # optional short title
            "description": row['question'],
            "difficulty": row['skill_level'],
            "starter_code": row['starter_code']
        }

        f.write(json.dumps({
            "instruction": instruction,
            "input": input_text,
            "output": output_data
        }) + "\n")

print("Conversion complete! Saved to", output_file)
```
